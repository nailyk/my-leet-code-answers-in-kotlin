```kotlin
fun hIndex(citations: IntArray): Int {
    val n = citations.size
    val paperCounts = IntArray(n + 1)  // from 0 to n

    // Count how many papers have i citations (cap at n)
    for (citation in citations) {
        if (citation >= n) {
            paperCounts[n]++
        } else {
            paperCounts[citation]++
        }
    }

    var total = 0
    // Go from high to low h-index possible
    for (i in n downTo 0) {
        total += paperCounts[i]
        if (total >= i) {
            return i
        }
    }

    return 0
}
```
