```kotlin
fun maxProfit(prices: IntArray): Int {

    var buyIndex = 0
    var sellIndex = 1
    var maxProfit = 0

    while (sellIndex < prices.size) {
        if (prices[sellIndex] < prices[buyIndex]) {
            buyIndex = sellIndex
        } else {
            maxProfit = max(maxProfit, prices[sellIndex] - prices[buyIndex])
        }
        sellIndex += 1
    }

    return maxProfit

}
```
