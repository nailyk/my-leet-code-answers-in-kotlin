```kotlin
import kotlin.random.Random

class RandomizedSet {
    private val indexMap = HashMap<Int, Int>() // val -> index
    private val values = ArrayList<Int>()

    fun insert(`val`: Int): Boolean {
        if (`val` in indexMap) return false
        indexMap[`val`] = values.size
        values.add(`val`)
        return true
    }

    fun remove(`val`: Int): Boolean {
        val index = indexMap[`val`] ?: return false
        val lastVal = values.last()

        // Move lastVal to index of val
        values[index] = lastVal
        indexMap[lastVal] = index

        // Remove last element
        values.removeAt(values.size - 1)
        indexMap.remove(`val`)

        return true
    }

    fun getRandom(): Int {
        return values[Random.nextInt(values.size)]
    }
}
```
