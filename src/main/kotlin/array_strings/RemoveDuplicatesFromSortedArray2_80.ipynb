```kotlin
fun removeDuplicates(nums: IntArray): Int {

    if (nums.size <= 2) return nums.size

    var k = 2
    for (i in 2 until nums.size) {
        if (nums[k-2] != nums[i]) {
            nums[k] = nums[i]
            k++
        }
    }
    return k
}
```

```kotlin
removeDuplicates(intArrayOf(1,1,1,2,2,3))
```
