Given an integer array `nums` of unique elements, return all possible subsets that can be made from the elements in `nums`.

The solution set must not contain duplicate subsets, and the subsets can be returned in any order.

Example 1:

Input:
```
nums = [1,2]
```
Output:
```
[[],[1],[2],[1,2]]
```
Example 2:

Input:
```
nums = [0]
```
Output:
```
[[],[0]]
```

````kotlin
fun subsets(nums: IntArray): List<List<Int>> {
    val result = mutableListOf<List<Int>>()

    fun dfs(index: Int, path: MutableList<Int>) {
        if (index == nums.size) {
            result.add(path.toList()) // copy
            return
        }

        // include nums[index]
        path.add(nums[index])
        dfs(index + 1, path)

        // exclude nums[index]
        path.removeAt(path.size - 1)
        dfs(index + 1, path)
    }

    dfs(0, mutableListOf())
    return result
}
````

````kotlin
// Example 1: two elements
println(subsets(intArrayOf(1, 2))) // [[], [1, 2], [1], [2]]

// Example 2: single element
println(subsets(intArrayOf(0))) // [[], [0]]

// Example 3: three elements
println(subsets(intArrayOf(1, 2, 3))) // [[], [1,2,3], [1,2], [1,3], [1], [2,3], [2], [3]]
````
