```kotlin
fun exist(board: Array<CharArray>, word: String): Boolean {

    fun dfs(r: Int, c: Int, index: Int): Boolean {
        if (index == word.length) return true
        if (
            r < 0 || r >= board.size ||
            c < 0 || c >= board[0].size ||
            board[r][c] != word[index]
        ) return false

        val temp = board[r][c]
        board[r][c] = '#'

        val found = dfs(r + 1, c, index + 1) ||
                dfs(r - 1, c, index + 1) ||
                dfs(r, c + 1, index + 1) ||
                dfs(r, c - 1, index + 1)

        board[r][c] = temp
        return found
    }

    val rows = board.size
    val cols = board[0].size

    for (row in 0..rows - 1) {
        for (col in 0..cols - 1) {
            if (board[row][col] == word[0]) {
                if (dfs(row, col, 0)) {
                    return true
                }
            }
        }
    }
    return false
}
```

```kotlin
exist(
    board = arrayOf(
        charArrayOf('A', 'B', 'C', 'E'),
        charArrayOf('S', 'F', 'C', 'S'),
        charArrayOf('A', 'D', 'E', 'E')
    ),
    word = "ASFCE"
)
```
