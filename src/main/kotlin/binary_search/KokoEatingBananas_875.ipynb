```kotlin
import java.util.Arrays

fun minEatingSpeed(piles: IntArray , h: Int): Int {

    var left = 1;
    var right = piles.max()

    // Binary search for minimum valid harvest rate
    while (left < right) {
        var mid = left + (right - left) / 2;
        if (timeTaken(piles, mid) > h) {
            // Rate too slow, need faster rate
            left = mid + 1;
        } else {
            // Rate is sufficient, try slower rate
            right = mid;
        }
    }

    return left;
}

private fun timeTaken(piles: IntArray, rate: Int): Int {
    var time = 0
    for (i in piles.indices) {
        time += (piles[i] + rate - 1) / rate // Ceiling division
    }
    return time
}

fun minHarvestRateBruteForce(piles: IntArray, h: Int): Int {
    var rate = 1
    while (timeTaken(piles, rate) > h) {
        rate++
    }
    return rate
}
```
