## Minimum Shipping Capacity

You are a logistics manager preparing to ship products from a warehouse. You have quantities of different product types that need to be packed into shipping boxes.

Each product type must be packed separately. Different product types cannot share the same box. All boxes have the same capacity. If a product has more items than fit in one box, multiple boxes are used.

You are given:
- An array of integers called quantities, where each value represents the number of items for a product type
- An integer called maxBoxes, representing the maximum number of boxes available

Your task is to determine the minimum box capacity required so that all products can be shipped using at most maxBoxes boxes.

### Example 1

Input:
```
quantities = [1, 2, 5, 9], maxBoxes = 6
```

Output:
5

Explanation:
With a box capacity of 5, the number of boxes needed is:
1 divided by 5 rounds up to 1
2 divided by 5 rounds up to 1
5 divided by 5 rounds up to 1
9 divided by 5 rounds up to 2

Total boxes used is 5, which is within the allowed limit.

6 divided by 4 rounds up to 2
7 divided by 4 rounds up to 2
11 divided by 4 rounds up to 3

Total boxes used is 8, which exactly meets the limit.
