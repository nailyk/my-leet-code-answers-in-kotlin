```kotlin
fun searchMatrix(matrix: Array<IntArray>, target: Int): Boolean {
    var row = 0
    var col = matrix[0].size - 1

    while (row < matrix.size && col >= 0) {
        val value = matrix[row][col]
        when {
            value == target -> return true
            value > target -> col-- // move left
            else -> row++           // move down
        }
    }

    return false
}
```

```kotlin
fun searchMatrix(matrix: Array<IntArray>, target: Int): Boolean {
    for (row in matrix) {
        if (binarySearch(row, target)) return true
    }
    return false
}

fun binarySearch(row: IntArray, target: Int): Boolean {
    var left = 0
    var right = row.size - 1

    while (left <= right) {
        val mid = left + (right - left) / 2
        when {
            row[mid] == target -> return true
            row[mid] < target -> left = mid + 1
            else -> right = mid - 1
        }
    }
    return false
}
```
