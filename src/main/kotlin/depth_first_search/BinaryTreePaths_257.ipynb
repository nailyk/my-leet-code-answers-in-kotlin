```kotlin
class TreeNode(var `val`: Int) {
    var left: TreeNode? = null
    var right: TreeNode? = null
}

fun binaryTreePaths(root: TreeNode?): List<String> {
    val result = mutableListOf<String>()
    if (root != null) dfs(root, "", result)
    return result
}

fun dfs(node: TreeNode, path: String, result: MutableList<String>) {
    val currentPath = if (path.isEmpty()) "${node.`val`}" else "$path->${node.`val`}"

    if (node.left == null && node.right == null) {
        // It's a leaf
        result.add(currentPath)
    } else {
        node.left?.let { dfs(it, currentPath, result) }
        node.right?.let { dfs(it, currentPath, result) }
    }
}
```
