```kotlin
fun rob(nums: IntArray): Int {
    if (nums.isEmpty()) return 0

    var prev = 0
    var curr = nums[0]

    for (i in 2..nums.size) {
        val take = prev + nums[i - 1]
        val skip = curr
        prev = curr
        curr = maxOf(take, skip)
    }

    return curr
}


fun robRecursive(treasure: IntArray): Int {
    val memo = mutableMapOf<Int, Int>()

    fun dp(i: Int): Int {
        if (i < 0) return 0
        if (memo.containsKey(i)) return memo[i]!!

        val robCurrent = dp(i - 2) + treasure[i]
        val skipCurrent = dp(i - 1)

        val result = maxOf(robCurrent, skipCurrent)
        memo[i] = result
        return result
    }

    return dp(treasure.lastIndex)
}
```
