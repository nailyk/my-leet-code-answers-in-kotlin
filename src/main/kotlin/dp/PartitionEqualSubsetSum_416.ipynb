```kotlin
fun canPartition(nums: IntArray): Boolean {
    val total = nums.sum()
    // If the total is odd, we cannot split it equally
    if (total % 2 != 0) return false

    val target = total / 2
    val dp = BooleanArray(target + 1)
    dp[0] = true  // sum 0 is always possible

    for (num in nums) {
        for (i in target downTo num) {
            if (dp[i - num]) dp[i] = true
        }
    }

    return dp[target]  // true if a subset with sum = target exists
}
```

```kotlin
canPartition(intArrayOf(3, 8, 11, 2))
```
