```kotlin
fun isHappy(n: Int): Boolean {
    var num = n
    val visit = mutableSetOf<Int>()

    while (num != 1 && !visit.contains(num)) {
        visit.add(num)
        num = getSumOfSquares(num)
    }

    return num == 1
}

fun getSumOfSquares(n: Int): Int {
    var sum = 0
    var number = n
    while (number > 0) {
        val digit = number % 10
        sum += digit * digit
        number /= 10
    }
    return sum
}
```

```kotlin
// optimal solution (linked list cycle)
fun isHappy(n: Int): Boolean {
    fun getSumOfSquares(num: Int): Int {
        var n = num
        var sum = 0
        while (n > 0) {
            val digit = n % 10
            sum += digit * digit
            n /= 10
        }
        return sum
    }

    var slow = n
    var fast = getSumOfSquares(n)

    while (fast != 1 && slow != fast) {
        slow = getSumOfSquares(slow)
        fast = getSumOfSquares(getSumOfSquares(fast))
    }

    return fast == 1
}
```
