```kotlin
data class ListNode(val value: Int) {
    var next: ListNode? = null
}

fun hasCycle(head: ListNode?): Boolean {

    var slow = head
    var fast = head

    while (fast != null && fast.next != null) {
        slow = slow?.next
        fast = fast.next?.next
        if (slow == fast) return true
    }
    return false
}
```

```kotlin
println(hasCycle(arrayToLinkedList(intArrayOf(3, 2, 0, 4))))

println(hasCycle(arrayToLinkedList(intArrayOf(3, 2, 0, 4), cycleIndex = 1)))

fun arrayToLinkedList(arr: IntArray, cycleIndex: Int = -1): ListNode? {
    if (arr.isEmpty()) return null
    val head = ListNode(arr[0])
    var current = head
    var nodeAtCycleIndex: ListNode? = null

    for (i in 1 until arr.size) {
        current.next = ListNode(arr[i])
        current = current.next!!
        if (i == cycleIndex) nodeAtCycleIndex = current
    }

    if (cycleIndex >= 0) current.next = nodeAtCycleIndex

    return head
}
```
