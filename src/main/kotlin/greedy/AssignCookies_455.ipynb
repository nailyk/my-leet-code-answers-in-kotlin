```kotlin
fun findContentChildren(g: IntArray, s: IntArray): Int {
    g.sort()
    s.sort()

    var i: Int = 0
    var j: Int = 0
    var count = 0

    while (i < g.size && j < s.size) {
        if (s[j] >= g[i]) {
            count += 1
            i += 1
        }
        j += 1
    }
    return count
}
```
