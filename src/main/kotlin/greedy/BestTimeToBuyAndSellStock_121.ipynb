```kotlin
fun maxProfit(prices: IntArray): Int {
    if (prices.isEmpty()) return 0

    var buy = prices[0]
    var maxProfit = 0

    for (sell in prices) {
        buy = minOf(buy, sell)
        maxProfit = maxOf(maxProfit, sell - buy)
    }

    return maxProfit
}
```

```kotlin
maxProfit(intArrayOf(2, 3, 5, 1, 10, 4, 3))
```
