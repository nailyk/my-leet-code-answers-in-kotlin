```kotlin
fun canCompleteCircuit(gas: IntArray, cost: IntArray): Int {
    var totalTank = 0
    var currTank = 0
    var startIndex = 0

    for (i in gas.indices) {
        val gain = gas[i] - cost[i]
        totalTank += gain
        currTank += gain

        if (currTank < 0) {
            startIndex = i + 1
            currTank = 0
        }
    }

    return if (totalTank >= 0) startIndex else -1
}
```

```kotlin
// Example 1: Solution exists at index 3
val gas1 = intArrayOf(1, 2, 3, 4, 5)
val cost1 = intArrayOf(3, 4, 5, 1, 2)
println(canCompleteCircuit(gas1, cost1)) // 3

// Example 2: No solution
val gas2 = intArrayOf(2, 3, 4)
val cost2 = intArrayOf(3, 4, 3)
println(canCompleteCircuit(gas2, cost2)) // -1
```
