```kotlin
fun canCompleteCircuit(gas: IntArray, cost: IntArray): Int {
    var totalTank = 0
    var currTank = 0
    var startIndex = 0

    for (i in gas.indices) {
        val gain = gas[i] - cost[i]
        totalTank += gain
        currTank += gain

        if (currTank < 0) {
            startIndex = i + 1
            currTank = 0
        }
    }

    return if (totalTank >= 0) startIndex else -1
}
```
