```kotlin
import java.util.PriorityQueue

fun kthLargest(nums: IntArray, k: Int): Int {
    if (nums.isEmpty()) {
        return -1
    }

    val heap = PriorityQueue<Int>()

    for (num in nums) {
        if (heap.size < k) {
            heap.add(num)
        } else if (num > heap.peek()) {
            heap.poll()
            heap.add(num)
        }
    }

    return heap.peek()
}
```

```kotlin
val nums = intArrayOf(3, 2, 1, 5, 6, 4)
val k = 4
println(kthLargest(nums, k)) // Output: 3
```

## Complexities

### Heap Solution - O(n log k)
- **Time**: O(n log k) - iterate through all elements (n) and maintain a heap of size k
- **Space**: O(k) - heap stores at most k elements

### Quick-Select - O(n) average, O(n²) worst case
- **Time**: O(n) average case, O(n²) worst case (poor pivot selection)
- **Space**: O(n) - creating new lists (less, greater) at each recursive call due to `drop()` and `filter()` operations

**Note**: The heap solution provides consistent performance with guaranteed O(n log k) complexity, while quick-select is faster on average but can degrade.

```kotlin
// COMPLEXITY on average : O(n)
fun findKthLargestQuickSelect(nums: IntArray, k: Int): Int {
    // convert kth largest to index from smallest
    val targetIndex = nums.size - k

    return quickSelect(nums, targetIndex)
}

fun quickSelect(list: IntArray, targetIndex: Int): Int {
    if (list.size <= 1) return list[0]

    val pivot = list[0]
    val less = list.drop(1).filter { it <= pivot }
    val greater = list.drop(1).filter { it > pivot }

    // The pivot's index in terms of the sorted order
    val pivotIndex = less.size

    // If the pivot is the target, return it
    return when {
        targetIndex == pivotIndex -> pivot
        targetIndex < pivotIndex -> quickSelect(less, targetIndex)
        else -> quickSelect(greater, targetIndex - pivotIndex - 1)
    }
}
```

```kotlin
val nums = intArrayOf(3, 2, 1, 5, 6, 4)
val k = 4
println(findKthLargestQuickSelect(nums, k)) // Output: 3
```
