Write a function to find the common free time for all employees from a list called `schedule`. Each employee's schedule is represented by a list of non-overlapping intervals sorted by start times. The function should return a list of finite, non-zero length intervals where all employees are free, also sorted in order

```kotlin
fun employeeFreeTime(schedule: Array<Array<IntArray>>): Array<IntArray> {
    // Step 1: Flatten all intervals
    val allIntervals = schedule.flatMap { it.asList() }.toMutableList()

    // Step 2: Sort intervals by start time
    allIntervals.sortBy { it[0] }

    val freeTimes = mutableListOf<IntArray>()

    // Step 3: Iterate from 1 -> n to find gaps
    for (i in 1 until allIntervals.size) {
        val prev = allIntervals[i - 1]
        val curr = allIntervals[i]

        if (curr[0] > prev[1]) {
            freeTimes.add(intArrayOf(prev[1], curr[0]))
        } else {
            // Extend prev[1] to handle overlapping intervals
            prev[1] = maxOf(prev[1], curr[1])
        }
    }

    return freeTimes.toTypedArray()
}
```

```kotlin
val schedule1 = arrayOf(
    arrayOf(intArrayOf(1, 3), intArrayOf(6, 7)),
    arrayOf(intArrayOf(2, 4))
)

val free1 = employeeFreeTime(schedule1)
free1.forEach { println(it.contentToString()) }
```
