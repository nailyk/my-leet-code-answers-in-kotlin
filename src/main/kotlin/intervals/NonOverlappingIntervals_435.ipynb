Write a function to return the minimum number of intervals that must be removed from a given array `intervals`, where `intervals[i]` consists of a starting point `starti` and an ending point `endi`, to ensure that the remaining intervals do not overlap.

```kotlin
fun eraseOverlapIntervals(intervals: Array<IntArray>): Int {

    if (intervals.size <= 1) return 0

    intervals.sortWith(compareBy { it[1] })

    var end = intervals[0][1]
    var count = 1

    for (i in 1 until intervals.size) {
        if (intervals[i][0] >= end) {
            end = intervals[i][1]
            count++
        }
    }

    return intervals.size - count
}
```

```kotlin
val intervals1 = arrayOf(
    intArrayOf(1, 2),
    intArrayOf(3, 4),
    intArrayOf(5, 6)
)

println(eraseOverlapIntervals(intervals1)) // 0
```

```kotlin
val intervals2 = arrayOf(
    intArrayOf(1, 3),
    intArrayOf(2, 4)
)

println(eraseOverlapIntervals(intervals2)) // 1
```

```kotlin
val intervals3 = arrayOf(
    intArrayOf(1, 2),
    intArrayOf(2, 3),
    intArrayOf(3, 4),
    intArrayOf(1, 3)
)

println(eraseOverlapIntervals(intervals3)) // 1
```

```kotlin
val intervals4 = arrayOf(
    intArrayOf(1, 5),
    intArrayOf(2, 6),
    intArrayOf(3, 7),
    intArrayOf(4, 8)
)

println(eraseOverlapIntervals(intervals4)) // 3
```
