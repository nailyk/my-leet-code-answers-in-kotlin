```kotlin
import linked_list.ListNode

fun addTwoNumbers(l1: ListNode?, l2: ListNode?): ListNode? {
    val myNewList = ListNode(0)

    var current = myNewList
    var p = l1
    var q = l2
    var carry = 0

    while (p != null || q != null) {
        val x = p?.`val` ?: 0
        val y = q?.`val` ?: 0
        val sum = x + y + carry

        carry = sum / 10
        current.next = ListNode(sum % 10)
        current = current.next!!

        if (p != null) p = p.next
        if (q != null) q = q.next
    }

    if (carry > 0) {
        current.next = ListNode(carry)
    }

    return myNewList.next
}
```
