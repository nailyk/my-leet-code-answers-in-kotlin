```kotlin
data class ListNode(val value: Int) {
    var next: ListNode? = null
}

fun isPalindrome(head: ListNode?): Boolean {

    if (head == null || head.next == null) return true

    // Find middle of the linked list using fast and slow pointers
    var slow = head
    var fast = head
    while (fast != null && fast.next != null) {
        slow = slow?.next
        fast = fast.next?.next
    }

    // Reverse second half of the list
    var prev: ListNode? = null
    while (slow != null) {
        val next = slow.next
        slow.next = prev
        prev = slow
        slow = next
    }

    // Check palindrome by comparing halves
    var left = head
    var right = prev
    while (right != null) {
        if (right.value != left?.value) return false
        left = left.next
        right = right.next
    }
    return true
}
```
