```kotlin
private fun isVowel(c: Char): Boolean =
    when (c) {
        'A', 'E', 'I', 'O', 'U',
        'a', 'e', 'i', 'o', 'u' -> true

        else -> false
    }

fun vowelStrings(word: String, queries: Array<IntArray>): IntArray {
    if (queries.isEmpty()) return IntArray(0)

    val prefix = IntArray(word.length + 1)

    for (i in 1..word.length) {
        prefix[i] = prefix[i - 1] + if (isVowel(word[i - 1])) 1 else 0
    }

    return IntArray(queries.size) { index ->
        val (start, end) = queries[index]
        prefix[end + 1] - prefix[start]
    }
}
```
