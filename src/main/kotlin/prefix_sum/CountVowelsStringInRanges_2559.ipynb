```kotlin
private fun isVowel(c: Char): Boolean =
    when (c) {
        'a', 'e', 'i', 'o', 'u' -> true
        else -> false
    }

fun vowelStrings(words: Array<String>, queries: Array<IntArray>): IntArray {
    if (queries.isEmpty()) return IntArray(0)

    val prefix = IntArray(words.size + 1)

    for (i in words.indices) {
        val word = words[i]
        val valid =
            isVowel(word.first()) && isVowel(word.last())
        prefix[i + 1] = prefix[i] + if (valid) 1 else 0
    }

    return IntArray(queries.size) { i ->
        val (l, r) = queries[i]
        prefix[r + 1] - prefix[l]
    }
}
```

```kotlin
vowelStrings(
    words = arrayOf("aba", "bcb", "ece", "aa", "e"),
    queries = arrayOf(intArrayOf(0, 2), intArrayOf(1, 4), intArrayOf(1, 1))
)
```
