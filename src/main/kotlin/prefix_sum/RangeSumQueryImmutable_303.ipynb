```kotlin
class NumArray(var nums: IntArray) {

    private val prefixSums = IntArray(nums.size)

    init {
        prefixSums[0] = nums[0]
        for (i in 1 until nums.size) {
            prefixSums[i] = prefixSums[i - 1] + nums[i]
        }
    }

    fun sumRange(left: Int, right: Int): Int {
        return if (left == 0) {
            prefixSums[right]
        } else {
            prefixSums[right] - prefixSums[left - 1]
        }
    }

}
```

```kotlin
NumArray(intArrayOf(3, 6, 7, 10, 1)).sumRange(1, 2)
```
