```kotlin
fun totalFruit(fruits: IntArray): Int {
    var start = 0
    val basket = HashMap<Int, Int>()
    var maxFruit = 0

    for (end in fruits.indices) {
        basket.put(fruits[end], basket.getOrDefault(fruits[end], 0) + 1)

        while (basket.size > 2) {
            val count = basket.getValue(fruits[start]) - 1
            if (count == 0) basket.remove(fruits[start])
            else basket.put(fruits[start], count)
            start++
        }

        maxFruit = maxOf(maxFruit, end - start + 1);

    }

    return maxFruit
}
```

```kotlin
totalFruit(intArrayOf(3,3,2,1,2,1,0))
```
