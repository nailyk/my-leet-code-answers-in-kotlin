```kotlin
fun characterReplacement(s: String, k: Int): Int {
    val charCount = mutableMapOf<Char, Int>()
    var maxFreq = 0
    var maxLength = 0
    var start = 0

    for (end in s.indices) {
        val endChar = s[end]
        charCount.put(endChar, (charCount.getOrDefault(endChar, 0)) + 1)
        maxFreq = maxOf(maxFreq, charCount.getValue(endChar))

        while ((end - start + 1) - maxFreq > k) { // (end - start + 1) - maxFreq <=> needed replacements count
            val leftChar = s[start]
            charCount.put(leftChar, charCount.getValue(leftChar) - 1)
            start++
        }

        maxLength = maxOf(maxLength, end - start + 1)
    }

    return maxLength

}
```

```kotlin
characterReplacement("BCBABCCCCA", k=2)
```
