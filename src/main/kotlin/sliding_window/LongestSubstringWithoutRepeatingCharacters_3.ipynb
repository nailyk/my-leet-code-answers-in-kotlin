```kotlin
fun longestSubstringWithoutRepeat(s: String): Int {
    var start = 0
    var maxLen = 0
    val seen = mutableSetOf<Char>()

    for (end in s.indices) {
        while (seen.contains(s[end])) {
            seen.remove(s[start])
            start += 1
        }
        seen.add(s[end])

        maxLen = maxOf(
            maxLen,
            end - start + 1 // or seen.size
       )

    }

    return maxLen
}

/*
    Faster optimization: jump start using last-seen indices
    Instead of shrinking the window one character at a time, you can jump start directly using a map.
 */
fun longestSubstringWithoutRepeatOptimized(s: String): Int {
    val lastSeen = mutableMapOf<Char, Int>()
    var start = 0
    var maxLen = 0

    for (end in s.indices) {
        val c = s[end]
        if (c in lastSeen) {
            start = maxOf(start, lastSeen.getValue(c) + 1)
        }
        lastSeen[c] = end
        maxLen = maxOf(maxLen, end - start + 1)
    }
    return maxLen
}
```
