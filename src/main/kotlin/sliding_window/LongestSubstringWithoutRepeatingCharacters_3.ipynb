```kotlin
fun lengthOfLongestSubstring(s: String): Int {
    val seen = mutableSetOf<Char>()
    var maxLen = 0
    var start = 0

    for (end in s.indices) {
        while (seen.contains(s[end])) {
            seen.remove(s[start])
            start++
        }
        seen.add(s[end])
        maxLen = maxOf(maxLen, end - start + 1)
    }

    return maxLen
}
```
