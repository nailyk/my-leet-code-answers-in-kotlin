```kotlin
fun maximumSubarraySum(nums: IntArray, k: Int): Long {
    var maxSum = 0L
    var currentSum = 0L
    var start = 0
    val freqMap = HashMap<Int, Int>()

    for (end in nums.indices) {
        currentSum += nums[end]
        freqMap[nums[end]] = freqMap.getOrDefault(nums[end], 0) + 1

        if (end - start + 1 == k) {
            if (freqMap.size == k) {
                maxSum = maxOf(maxSum, currentSum)
            }

            currentSum -= nums[start].toLong()
            freqMap[nums[start]] = freqMap.getValue(nums[start]) - 1
            if (freqMap[nums[start]] == 0) {
                freqMap.remove(nums[start])
            }
            start++
        }
    }

    return maxSum
}
```

```kotlin
maximumSubarraySum(intArrayOf(2, 1, 5, 1, 3, 2), k = 3)
```
