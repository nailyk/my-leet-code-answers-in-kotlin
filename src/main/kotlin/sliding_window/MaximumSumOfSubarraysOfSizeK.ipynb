```kotlin
fun maxSum(nums: IntArray, k: Int): Int {

    var maxSum = 0
    var windowSum = 0
    var start = 0

    for (end in nums.indices) {
        windowSum += nums[end]
        if (end - start + 1 == k) {
            maxSum = maxOf(maxSum, windowSum)
            windowSum -= nums[start]
            start += 1
        }
    }
    return maxSum

}
```

```kotlin
maxSum(intArrayOf(2, 1, 5, 1, 3, 2), k = 3)
```
