```kotlin
fun simplifyPath(path: String): String {
    val parts = path.split("/")
    val stack = ArrayDeque<String>()

    for (part in parts) {
        when {
            part.isEmpty() || part == "." -> continue
            part == ".." -> if (stack.isNotEmpty()) stack.removeLast()
            else -> stack.addLast(part)
        }
    }

    return "/" + stack.joinToString("/")
}
```
