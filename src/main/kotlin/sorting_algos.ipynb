<h1>Binary search</h1>

```kotlin
fun binarySearch(nums: IntArray, target: Int): Int {
    var low = 0
    var high = nums.size - 1
    while (low <= high) {
        val mid = low + (high - low) / 2 // not doing (low + high) / 2 to avoid potential overflow !
        when {
            nums[mid] == target -> return mid
            nums[mid] < target -> low = mid + 1
            else -> high = mid - 1
        }
    }
    return -1
}
```

```kotlin
println(binarySearch(intArrayOf(3,5,2,1,6,3,10,9,12), target = 10))
```

<h1> Selection sort

```kotlin
fun selectionSort(arr: IntArray) {
    val n = arr.size
    for (i in 0 until n - 1) {
        // Find the minimum element in unsorted part of array
        var minIndex = i
        for (j in i + 1 until n) {
            if (arr[j] < arr[minIndex]) {
                minIndex = j
            }
        }

        // Swap the found minimum element with the first element of the unsorted part
        if (minIndex != i) {
            val temp = arr[i]
            arr[i] = arr[minIndex]
            arr[minIndex] = temp
        }
    }
}
```

```kotlin
val array = intArrayOf(64, 25, 12, 22, 11)
println("Original Array: ${array.joinToString(", ")}")

selectionSort(array)

println("Sorted Array: ${array.joinToString(", ")}")
```

<h1>Quick sort</h1>

```kotlin
fun quicksort(list: List<Int>): List<Int> {
    if (list.size < 2) return list

    val pivot = list[0]
    val less = list.drop(1).filter { it <= pivot }
    val greater = list.drop(1).filter { it > pivot }

    return quicksort(less) + pivot + quicksort(greater)
}
```

```kotlin
val nums = listOf(10, 5, 2, 3, 7, 6, 8)
val sorted = quicksort(nums)
println("Sorted: $sorted")
```

<h1>BFS</h1>


```kotlin
import java.util.*

fun bfs(graph: Map<Int, List<Int>>, start: Int) {
    val visited = mutableSetOf<Int>()
    val queue: Queue<Int> = LinkedList()

    visited.add(start)
    queue.add(start)

    while (queue.isNotEmpty()) {
        val node = queue.poll()
        println("Visited: $node")

        for (neighbor in graph[node] ?: emptyList()) {
            if (neighbor !in visited) {
                visited.add(neighbor)
                queue.add(neighbor)
            }
        }
    }
}
```
