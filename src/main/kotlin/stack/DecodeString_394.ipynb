```kotlin
fun decodeString(s: String): String {
    val stringStack = ArrayDeque<StringBuilder>()
    val numberStack = ArrayDeque<Int>()
    var currentString = StringBuilder()
    var currentNumber = 0

    for (c in s) {
        when {
            c.isDigit() -> currentNumber = currentNumber * 10 + c.digitToInt()
            c == '[' -> {
                stringStack.addLast(currentString)
                numberStack.addLast(currentNumber)
                currentString = StringBuilder()
                currentNumber = 0
            }
            c == ']' -> {
                val repeatTimes = numberStack.removeLast()
                val prevString = stringStack.removeLast()
                repeat(repeatTimes) { prevString.append(currentString) }
                currentString = prevString
            }
            else -> currentString.append(c)
        }
    }
    return currentString.toString()
}
```

```kotlin
decodeString("4[a2[b]]")
```
