```kotlin
fun nextGreaterElement(nums1: IntArray, nums2: IntArray): IntArray {
    val nextGreaterMap = mutableMapOf<Int, Int>()
    val stack = ArrayDeque<Int>()

    for (num in nums2) {
        // Maintain decreasing stack
        while (stack.isNotEmpty() && num > stack.last()) {
            val smaller = stack.removeLast()
            nextGreaterMap[smaller] = num
        }
        stack.addLast(num)
    }

    // Remaining numbers have no greater element
    for (num in stack) {
        nextGreaterMap[num] = -1
    }

    // Build result for nums1 using the map
    return nums1.map { nextGreaterMap[it] ?: -1 }.toIntArray()
}
```
