```kotlin
fun maxArea(height: IntArray): Int {
    var left = 0
    var right = height.lastIndex
    var maxArea = 0

    while (left < right) {
        val h = minOf(height[left], height[right])
        val w = right - left
        maxArea = maxOf(maxArea, h * w)

        if (height[left] < height[right]) {
            left++
        } else {
            right--
        }
    }

    return maxArea
}
```
