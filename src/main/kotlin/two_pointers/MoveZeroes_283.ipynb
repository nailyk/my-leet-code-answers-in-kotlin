```kotlin
fun moveZeroes(nums: IntArray) {
    var nextNonZero = 0

    for (i in nums.indices) {
        if (nums[i] != 0) {
            val temp = nums[nextNonZero]
            nums[nextNonZero] = nums[i]
            nums[i] = temp
            nextNonZero++
        }
    }
}
```

```kotlin
val nums = intArrayOf(0, 1, 2, 1, 5)
moveZeroes(nums)

println(nums.joinToString())
```
