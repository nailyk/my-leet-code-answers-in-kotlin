Write a function to sort a given integer array `nums` in-place (and without the built-in sort function), where the array contains `n` integers that are either 0, 1, and 2 and represent the colors red, white, and blue. Arrange the objects so that same-colored ones are adjacent, in the order of red, white, and blue (0, 1, 2).

Input:
```
nums = [2,1,2,0,1,0,1,0,1]
```

Output:
```
[0,0,0,1,1,1,1,2,2]
```

````kotlin
fun sortColors(nums: IntArray): Unit {
    // [0, left - 1] are all 0s
    // [left, i - 1] are all 1s
    // [i, right] are unexplored
    // [right + 1, nums.length - 1] are all 2s

    var left = 0
    var right = nums.size - 1
    var i = 0
    while (i <= right) {
        if (nums[i] == 0) {
            var temp = nums[i]
            nums[i] = nums[left]
            nums[left] = temp
            left++
            i++
        } else if (nums[i] == 1) {
            i++
        } else {
            var temp = nums[i]
            nums[i] = nums[right]
            nums[right] = temp
            right--
        }
    }
}
````

````kotlin
// Example 1: mixed colors
val ex1 = intArrayOf(2, 1, 2, 0, 1, 0, 1, 0, 1)
sortColors(ex1)
println(ex1.toList()) // [0, 0, 0, 1, 1, 1, 1, 2, 2]

// Example 2: already sorted
val ex2 = intArrayOf(0, 0, 1, 1, 2, 2)
sortColors(ex2)
println(ex2.toList()) // [0, 0, 1, 1, 2, 2]

// Example 3: reverse sorted
val ex3 = intArrayOf(2, 2, 1, 1, 0, 0)
sortColors(ex3)
println(ex3.toList()) // [0, 0, 1, 1, 2, 2]

// Example 4: single element
val ex4 = intArrayOf(1)
sortColors(ex4)
println(ex4.toList()) // [1]
````
